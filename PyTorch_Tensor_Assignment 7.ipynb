## Mathematical Operations

### 1. Scalar Operations

```python
x = torch.rand(2,2)
x
```

```python
x+5
x-1
x*4
x/2
x**3
```

### 2. Element Wise Operations

```python
a=torch.rand(2,2)
b=torch.rand(2,2)
print(a)
print(b)
```

```python
a+b
a-b
a*b
a/b
```

```python
c=torch.tensor([-5,3,-8,10])
torch.abs(c)
```

```python
d=torch.tensor([1.2,4.8,6.1,2.9])
torch.round(d)
```

```python
torch.ceil(d)
torch.floor(d)
torch.clamp(d,min=2,max=5)
```

### 3. Reduction Operations

```python
e=torch.randint(1,20,(3,3),dtype=torch.float32)
e
```

```python
torch.sum(e)
torch.mean(e)
torch.max(e)
torch.min(e)
```

```python
torch.std(e)
torch.var(e)
torch.argmax(e)
```

### 4. Matrix Operations

```python
m1=torch.tensor([[1.,2.],[3.,4.]])
m2=torch.tensor([[5.,6.],[7.,8.]])
```

```python
torch.matmul(m1,m2)
```

```python
m1.T
```

### 5. Comparison Operations

```python
x=torch.tensor([2,4,6,8])
y=torch.tensor([1,4,7,8])
```

```python
x==y
x>y
x<y
```

### 6. Special Functions

```python
x=torch.tensor([1.0,2.0,3.0])
torch.sqrt(x)
```

```python
torch.exp(x)
```

```python
torch.log(x)
```
